# Connect sqlite database

Raw data paths pattern: `PROJ_ROOT/data/raw/briefings/YYYY/YYYY-MM/YYYY-MM-DD.json`

Raw data fields:

- `date` --> unique key in the database
- `url` # `page_url = f"{BASE_URL}/{date_string}.html"`
- `status` 
- `http_status` --> nullable, SQLite column types/constraints must allow NULLs rather than empty string
- `html` # raw HTMl --> need a parsing step first
- `error` --> nullable, SQLite column types/constraints must allow NULLs rather than empty string
- `downloaded_at` # `datetime.now(timezone.utc).isoformat(timespec="seconds")`

The module should:

- allows to query, insert, update, and delete data from the database passing a briefing_id.
- Implement bulk load method or function to insert data from the JSON files.
- Implement preprocessing steps split briefings into sections and insert them into the database
- Include unit tests

Schema from Franz:

-- SQLite schema for UN LLM Analysis

PRAGMA foreign_keys = ON;

CREATE TABLE briefings (
    briefing_id INTEGER PRIMARY KEY,
    briefing_date TEXT NOT NULL UNIQUE,
    url TEXT NOT NULL UNIQUE,
    speaker_name TEXT NOT NULL,
    speaker_role TEXT NOT NULL,
    full_text TEXT NOT NULL,
    downloaded_at TEXT NOT NULL,
    processed_at TEXT NOT NULL
);


CREATE TABLE sections (
    section_id INTEGER PRIMARY KEY,
    briefing_id INTEGER NOT NULL,
    section_order INTEGER NOT NULL,
    title TEXT NOT NULL,
    text TEXT NOT NULL,

    FOREIGN KEY (briefing_id)
        REFERENCES briefings(briefing_id)
        ON DELETE CASCADE,

    UNIQUE (briefing_id, section_order)
);
