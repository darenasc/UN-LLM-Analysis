````python
import ollama
````

````python
# extracts from https://www.un.org/sg/en/content/noon-briefing-highlight?date=2026-08-14
section_texts = [
    """The UN Children’s Fund (UNICEF) today said that more than 2.6 million girls in Afghanistan have been denied a secondary education in the five years since the de facto authorities took power. UNICEF warns that girls who are out of school face heightened protection risks, including child labour, abuse, early marriage, and gender-based violence.  

Despite the restrictions, UNICEF continues to support learning in Afghanistan. """,
    """In the Occupied Palestinian Territory, in the West Bank, UN humanitarian partners warn that aid work is being delayed and limited by a network of over 900 physical obstacles – including checkpoints, gates, roadblocks and other restrictions. This hinders humanitarian access and leave people in need underserved. 

Today, aid and municipal workers have been struggling to secure access to three Palestinian families in Qusra village near Nablus. These families have been trapped there since Tuesday, following the establishment of a settlement outpost next to their homes and the tightening of access restrictions in the area. 

Alongside municipal workers, UN partners have been trying to deliver food, water, milk formula, medicines and hygiene items to those trapped inside, but such attempts have been unsuccessful so far. UN partners warn that efforts to protect Palestinians have been ineffective, with settlers reportedly reinstalling a tent that Israeli forces had dismantled yesterday. 

Meanwhile, in the northern West Bank, where the humanitarian situation remains particularly severe, child protection partners have provided vital assistance, including psychosocial support, case management, and services for children with disabilities, to nearly 30,000 children and 15,000 adults since the start of this year. 

Turning to Gaza. The Office for the Coordination of Humanitarian Affairs (OCHA) said that humanitarians continue to provide vital aid to people in Gaza, despite persistent insecurity and restrictions. 

Last week, the UN and its partners mobilized rapid support for people who had lost their shelters or were displaced again due to attacks, domestic fires and other incidents. 

In more than a dozen reported incidents, people were displaced and lost their belongings due to attacks, impacting nearly 290 households. Two households were affected by domestic fires, which are often linked to unsafe cooking in shelters in the absence of proper energy sources. Another five households were affected by the placement of new yellow cement blocks, which often triggers displacement in Gaza by signaling the expansion of areas where access is restricted. 

Most of last week’s appeals for help come from northern Gaza and Deir al Balah, with the UN and its partners tailoring the response to meet families’ specific needs. 

Also last week, UN partners report that more than 300,000 children continued to participate in summer learning programmes they support. To address the impact of extreme heat in Gaza, UN education colleagues installed shade structures in 10 temporary learning spaces.""",
    """From Ukraine, the Office for the Coordination of Humanitarian Affairs (OCHA) report that they continue to deliver humanitarian assistance to front-line areas that are hard to reach, at a time when the security situation is worsening across the country, driving needs even higher. Yesterday and today, OCHA, with support from partners leading on logistics, facilitated two humanitarian convoys to front-line communities in the Kherson and Kharkiv regions, delivering hygiene supplies, first-aid kits and food.""",
]
````

````python
# Use the generate function for a one-off prompt

for section in section_texts:
    result = ollama.generate(
        model="qwen2.5vl:3b",
        prompt=f"""Generate a list with the names of countries or territories mentioned in the following context: 
        
        Context:
        
        {section}

    Answer only based on the context passed. If you don't know if it is a country or territory, reply with 'None'.

    The response should be in a list. For example: ['Germany', 'UK', 'Occupied Palestinian Territory'].
    """,
    )
    print(result["response"])
    print("*" * 20)
````
